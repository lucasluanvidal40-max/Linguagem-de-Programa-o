```python
import matplotlib.pyplot as plt

class Livro:
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidade = quantidade

biblioteca = []

def cadastrar_livro(titulo, autor, genero, quantidade):
    novo_livro = Livro(titulo, autor, genero, quantidade)
    biblioteca.append(novo_livro)
    print(f'Sucesso: Livro "{titulo}" cadastrado!')

def listar_livros():
    if not biblioteca:
        print("A biblioteca está vazia no momento.")
        return

    print("\n--- Lista de Livros Disponíveis ---")
    for livro in biblioteca:
        print(f'Título: {livro.titulo} | Autor: {livro.autor} | Gênero: {livro.genero} | Estoque: {livro.quantidade}')
    print("-----------------------------------")

def buscar_livro(titulo_busca):
    encontrado = False
    print(f'\nBuscando por: "{titulo_busca}"...')
    for livro in biblioteca:
        if livro.titulo.lower() == titulo_busca.lower():
            print(f'=> ENCONTRADO: Título: {livro.titulo} | Autor: {livro.autor} | Gênero: {livro.genero} | Estoque: {livro.quantidade}')
            encontrado = True
            break

    if not encontrado:
        print("=> Livro não encontrado na biblioteca.")

def gerar_grafico_generos():
    if not biblioteca:
        print("\nNão há dados suficientes para gerar o gráfico.")
        return

    contagem_generos = {}
    for livro in biblioteca:
        if livro.genero in contagem_generos:
            contagem_generos[livro.genero] += livro.quantidade
        else:
            contagem_generos[livro.genero] = livro.quantidade

    generos = list(contagem_generos.keys())
    quantidades = list(contagem_generos.values())

    plt.figure(figsize=(8, 5))
    plt.bar(generos, quantidades, color=['#4C72B0', '#55A868', '#C44E52'])
    plt.title('Quantidade de Livros por Gênero', fontsize=14)
    plt.xlabel('Gênero', fontsize=12)
    plt.ylabel('Quantidade em Estoque', fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    plt.show()


cadastrar_livro("O Senhor dos Anéis", "J.R.R. Tolkien", "Fantasia", 10)
cadastrar_livro("O Hobbit", "J.R.R. Tolkien", "Fantasia", 5)
cadastrar_livro("1984", "George Orwell", "Ficção Científica", 8)
cadastrar_livro("Duna", "Frank Herbert", "Ficção Científica", 4)
cadastrar_livro("Dom Casmurro", "Machado de Assis", "Romance", 6)

listar_livros()

buscar_livro("1984")
buscar_livro("Harry Potter")

print("\nGerando gráfico de gêneros...")
gerar_grafico_generos()
```
